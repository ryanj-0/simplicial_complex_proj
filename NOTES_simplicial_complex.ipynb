# Partially Ordered Set (PoSet)

- Set, $S$; Binary relation $R$, $\preceq$
  - (S, $\preceq$)
  - $(a,b) \in R$
  - _Partial_ $\implies$ not all pairs are comparable.
- all elements $x, y, z \in S$:
    - $x \preceq x$ (Reflexive)
    - $x \preceq y$ and $y \preceq x$ $\implies x = y$ (Anti-Symmetry)
    - $x \preceq y$ and $y \preceq z$ $\implies x \preceq z$ (Transitivity)

**Example (below)**

Let R be the relation $\subseteq$, so we then have (S,$\subseteq$)

Note: ((1,), (2,)) is not in R. Neither element is subset of each other so 
it will not be in R. This is what partial ordering allows.

```python
p = PoSet(
    S = {(1,), (2,), (1,2)}, 
    R = {((1,),(1,)), ((2,),(2,)), ((1,2),(1,2)), ((1,),(1,2)), ((2,),(1,2))}
)

print("The relation R on S is the set:")
print(*p.format_relation(), sep="\n")
```

# Simplicial Complex (Abstract)



### Why a subclass?
- We still have a Poset if we just had a collection of sets with subset
  - what makes it a subclass is the requirement of all non-empty subsets

### How we can talk about Simplicial Complexs:

#### Helper
- n-simplex
- Faces
- Maximal
- Hasse Diagrams

#### Methods
- Dimension
- Connected
- Boundary
- Euler Characteristic
- Surface
- Orientable


```python
# Two Triangles and a Tail
two_triangles = SimplicialComplex([(1,2,3),(2,3,4),(3,5)])
print(*sorted(two_triangles.S, key=lambda f: len(f)), sep="\n")
two_triangles.s_complex_graph()
```

```python
# Simple (Complex with 2, 1, and 0-simplex)
test_complex = SimplicialComplex([{"a","b","c"}, {"c","d"}, {"e"}])
print(f"Elements in complex: {len(test_complex.S)}")
print(f"Relation set: {len(test_complex.R)}")
print(test_complex)

print("")

print(f"Dim: {test_complex.dim()}")
```

```python
print(f"Boundary of abc: {test_complex.boundary_k("abc")}")


print("")
test_complex.draw_hasse()
```
